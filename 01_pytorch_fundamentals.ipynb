```python
import torch
```

```python
x=torch.arange(1,10)
```

```python
x,x.shape
```

```python
#Add an extra dimension
x_reshaped=x.reshape(1,9)
x_reshaped,x_reshaped.shape
```

```python
#Change the view
z=x.view(1,9)
z,z.shape
```

```python
#Changing z changes x (because a view of tensor share the same memory as the original input)
z[:,0]=5
z,x
```

```python
#Stack tensor over top of each other

x_stacked=torch.stack([x,x,x,x],dim=0)
x_stacked
```

```python
#Stack tensor over top of each other

x_stacked=torch.stack([x,x,x,x],dim=1)
x_stacked
```

```python
#torch.squeeze() -remove all single dimension from large tensor
squeeze=torch.squeeze(x)
squeeze,squeeze.shape
```

```python
x.shape
```

```python
#torch.squeeze() -remove all single dimension from large tensor
print(f'Previous tensor:{x}')
print(f'Previous shape:{x.shape}')
x_squeezed=x.squeeze()
print(f'New tensor:{x_squeezed}')
print(f'New shape:{x_squeezed.shape}')
```

```python
#torch.unsqueeze()-adds a single dimension to  a target tensor at a specific dimension
print(f'Previous tensor:{x_squeezed}')
print(f'Previous shape:{x_squeezed.shape}')
x_unsqueezed=x_squeezed.unsqueeze(dim=0)
print(f'New tensor:{x_unsqueezed}')
print(f'New shape:{x_unsqueezed.shape}')
```

```python
#torch.unsqueeze()-adds a single dimension to  a target tensor at a specific dimension
print(f'Previous tensor:{x_squeezed}')
print(f'Previous shape:{x_squeezed.shape}')
x_unsqueezed=x_squeezed.unsqueeze(dim=1)
print(f'New tensor:{x_unsqueezed}')
print(f'New shape:{x_unsqueezed.shape}')
```
